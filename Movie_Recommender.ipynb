We have to get the output of the previous cell into this format ['Action', 'Adventure', 'Fantasy', 'Science Fiction']

For that, we should convert this '[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]' into a list

```python
#We use ast module for this purpose and literal eval as the sub module from the ast module
import ast
ast.literal_eval('[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]')
```

Now, this is in the lsit format 

```python
def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
```

Now, we have to get top 3 actors as cast for the film, so, we need the access for first three dictionaries

```python
def convert3(obj):
    L = []
    counter = 0
    for i in ast.literal_eval(obj):
        if counter != 3:
            L.append(i['name'])
            counter = counter + 1
        else: 
            break
    return L
```

To get the dir. of the film by iterating through the department in every dictionary

```python
def fetch_director(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            L.append(i['name'])
            break;
    return L
```

We have to use a technique called stemming, Where the words like loved, loving, love then we will get the output as love, love, love

Because, now, in the data we have action, actions as two words which might disrupt the model with a partial redundancy. We have to get rid of this

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []

    for i in text.split():
        y.append(ps.stem(i))

    return " ".join(y)
```

Now, as every movie is in the form of a vector. We have to calculate the distance of one movie's vector with every other movie's vector. If the distance of the vectors from a movie is low,  then those movies are generally recommended

For calculating the distance, we are using the cosine distance 

Cosine distance is inversely proportional to the similarity between the movies 

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
similarity
```

```python
similarity.shape
```

This similarity word gives us the cosine distance of a vector from every other vector

```python
similarity[0]
```

Now, this array gives us the similarity of 1st movie with every other movie, but we need to arrange these in a sorted order, by not losing the actual index of the array. Because, if we sort the array then the 1st index of the array(2nd place) will be occupied by some other movie which is not present actually in the index 2.
Suppose Inception is 1st movie
Predestination is 2nd movie, then after sorting there might be the possibility of occuring of another thriller movie in the mid like fight club. 

So, we should use enumerate function

```python
sorted(list(enumerate(similarity[0])), reverse = True)#This sorted function sorts the data based on the index value
```

But, we have a need to change the data based on the similarity, Therefore:-

```python
sorted(list(enumerate(similarity[0])), reverse = True, key = lambda x:x[1])[1:6]
```

```python
new_df['title'][1216]
```

```python
def recommend(movie):
    movie_index = new_df[new_df['title'] == movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(sorted(list(enumerate(distances)), reverse = True, key = lambda x:x[1])[1:6])

    for i in movies_list:
        print(new_df.iloc[i[0]].title)
```

```python
recommend("Avatar")
```
